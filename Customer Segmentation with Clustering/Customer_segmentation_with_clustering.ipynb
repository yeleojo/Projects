Since we have a transnational data set, which implies customers from different continents, several metrics are important when performing customer segmentation for target marketing. From a marketing perspective, the following five metrics help to understand the nuance of the customer base, buying behaviour, preferences, and value to the business.

CLV helps to prioritise marketing efforts and resources as it focuses on customers who are expected to bring the most value over time. Therefore, retaining high-value customers.
- The **average unit cost** indicates if the customer prefers low cost or high cost items. This is related to the profitability of purchases. Customers buying products with a higher average unit cost price should be targeted differently. Customer segmentation assists in identifying these customers.

## Feature Engineering

```python
# creating a new feature for frequency
# frequency indicates how often a customer purchased over the given period of time covered by the dataset
data['Frequency'] = data.groupby('Customer ID')['Order ID'].transform('nunique')

```

```python
# View the data
data.head()
```

```python
# Calculate and create new feature for recency, representing time since last order date

data['Order_Date'] = pd.to_datetime(data['Order_Date'], format='%d%b%Y')
data['Recency'] = (pd.to_datetime('today') - data['Order_Date']).dt.days

# View the data
data
```

```python
# Create new feature for CLV, total value customer contributes to the business over the course of the relationship.
# convert the Total revenue column to number format
data['Total Revenue'] = data['Total Revenue'].astype(str).str.replace('$', '').str.replace(',', '').astype(float)

# Rename Total revenue column to CLV
data.rename(columns={'Total Revenue': 'CLV'}, inplace=True)

# View the data
data.head()
```

```python
# Create Average unit cost feature. This shows if a customer prefers low cst or high cost items
# convert unit cost column to number format
data['Unit Cost'] = data['Unit Cost'].astype(str).str.replace('$', '').str.replace(',', '').astype(float)

# Create Average unit cost feature
#data['Unit Cost'] = data['Unit Cost'].apply(currency_to_float)

# rename column to Avg_Unit_Cost
data.rename(columns={'Unit Cost': 'Avg_Unit_Cost'}, inplace=True)

# View the data
data
```

```python
# Create customer age feature by converting the DOB to datetime

data['Customer_BirthDate'] = pd.to_datetime(data['Customer_BirthDate'], format='%d%b%Y')
data['Customer_Age'] = (pd.to_datetime('today') - data['Customer_BirthDate']).dt.days // 365.25

# View the data
data.head()
```

```python
# Aggregate to get one record by customer, grouping by customer ID
# Create new customer_df dataframe

customer_df = data.groupby('Customer ID').agg(
    Recency=('Recency', 'max'),
    Frequency=('Order ID', 'nunique'),
    CLV=('CLV', 'sum'),
    Average_Unit_Cost=('Avg_Unit_Cost', 'mean'),
    Customer_Age=('Customer_Age', 'max')
).reset_index()

# view the data
print(customer_df.head())

# view the shape
customer_df.shape
```

# Clustering with ML models

```python
# perform k-means clustering with scikit-learn
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score

kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
customer_df['Cluster'] = kmeans.fit_predict(customer_df)

#Extracting centroids
centroids = kmeans.cluster_centers_


#plottimg the clusters and centroids using plt.scatter
for cluster_id in range(5):
    cluster_data = customer_df[customer_df['Cluster'] == cluster_id]
    plt.scatter(cluster_data['Recency'], cluster_data['Frequency'], label=f'Cluster {cluster_id}')
    plt.scatter(centroids[cluster_id, 0], centroids[cluster_id, 1], marker='X', color='red', s=100, label=f'Centroid {cluster_id}')

plt.xlabel('Recency')
plt.ylabel('Frequency')
plt.title('K-Means Clustering with Centroids')
plt.legend()
plt.show()

# print final centroids
print("Final Centroids:")
print(centroids)
```

```python


# Elbow method: try different k.
wcss=[]
for i in range(1,11):
    kmeans=KMeans(n_clusters=i, init='k-means++',random_state=0, n_init = 10)
    kmeans.fit(customer_df)
    wcss.append(kmeans.inertia_)

plt.plot(range(1,11),wcss)
plt.title('The Elbow Method')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.show()
```

Using the elbow method, the elbow value appears to be at k = 4

## Silhouette Score method

```python
# silhouette score method for different values of n_clusters
#from sklearn.metrics import silhouette_score
#from sklearn.cluster import KMeans

# range_n_clusters = [2, 3, 4, 5, 6]



```

```python
from sklearn.metrics import silhouette_samples
from sklearn.metrics import silhouette_score
from sklearn.cluster import KMeans
import matplotlib.cm as cm # Import colormap

range_n_clusters = [2, 3, 4, 5, 6]


for n_clusters in range_n_clusters:
    # Create a subplot with 1 row and 2 columns.
    fig, (ax1, ax2) = plt.subplots(1, 2)
    fig.set_size_inches(18, 7)

    # The 1st subplot is the silhouette plot.
    # The silhouette coefficient can range from -1, 1 but in this example all
    # lie within [-0.1, 1].
    ax1.set_xlim([-0.1, 1])
    # The (n_clusters+1)*10 is for inserting blank space between silhouette
    # plots of individual clusters, to demarcate them clearly.
    ax1.set_ylim([0, len(customer_df) + (n_clusters + 1) * 10])

    # Initialize the clusterer with n_clusters value and a random generator
    # seed of 10 for reproducibility.
    clusterer = KMeans(n_clusters=n_clusters, random_state=10, n_init = 10)
    cluster_labels = clusterer.fit_predict(customer_df)

    # The silhouette_score gives the average value for all the samples.
    # This gives a perspective into the density and separation of the formed
    # clusters
    silhouette_avg = silhouette_score(customer_df, cluster_labels)
    print("For n_clusters =", n_clusters,
          "The average silhouette_score is :", silhouette_avg)

    # Compute the silhouette scores for each sample
    sample_silhouette_values = silhouette_samples(customer_df, cluster_labels)

    y_lower = 10
    for i in range(n_clusters):
        # Aggregate the silhouette scores for samples belonging to
        # cluster i, and sort them.
        ith_cluster_silhouette_values = \
            sample_silhouette_values[cluster_labels == i]

        ith_cluster_silhouette_values.sort()

        size_cluster_i = ith_cluster_silhouette_values.shape[0]
        y_upper = y_lower + size_cluster_i

        color = cm.nipy_spectral(float(i) / n_clusters)
        ax1.fill_betweenx(np.arange(y_lower, y_upper),
                          0, ith_cluster_silhouette_values,
                          facecolor=color, edgecolor=color, alpha=0.7)

        # Label the silhouette plots with their cluster numbers at the middle.
        ax1.text(-0.05, y_lower + 0.5 * size_cluster_i, str(i))

        # Compute the new y_lower for next plot.
        y_lower = y_upper + 10  # 10 for the 0 samples.

    ax1.set_title("The silhouette plot for the various clusters.")
    ax1.set_xlabel("The silhouette coefficient values")
    ax1.set_ylabel("Cluster label")

    ax1.set_yticks([])  # Clear the yaxis labels / ticks.
    ax1.set_xticks([-0.1, 0, 0.2, 0.4, 0.6, 0.8, 1])

    # 2nd Plot showing the actual clusters formed.
    colors = cm.nipy_spectral(cluster_labels.astype(float) / n_clusters)
    ax2.scatter(customer_df.iloc[:, 0], customer_df.iloc[:, 1], marker='.', s=30, lw=0, alpha=0.7,
                c=colors, edgecolor='k')

    # Labeling the clusters.
    centers = clusterer.cluster_centers_
    # Draw white circles at cluster centers.
    ax2.scatter(centers[:, 0], centers[:, 1], marker='o',
                c="white", alpha=1, s=200, edgecolor='k')

    for i, c in enumerate(centers):
        ax2.scatter(c[0], c[1], marker='$%d$' % i, alpha=1,
                    s=50, edgecolor='k')

    ax2.set_title("The visualization of the clustered data.")
    ax2.set_xlabel("Feature space for the 1st feature")
    ax2.set_ylabel("Feature space for the 2nd feature")

    plt.suptitle(("Silhouette analysis for KMeans clustering on sample data "
                  "with n_clusters = %d" % n_clusters),
                 fontsize=14, fontweight='bold')

plt.show()
```

From the results, n_clusters of 2 has the highest average silhouette score (0.409), suggesting it might be a better choice than 3, 4, 5, or 6 based on this metric. Considering both the elbow method and the silhouette score, a choice between k=2, k=3 or k=4 seems reasonable.
